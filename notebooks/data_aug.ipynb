```python
import json
import random
from datasets import Dataset
```

```python
dataset_path = "../data/synthetic_dataset_v1.json"

with open(dataset_path, "r") as f:
    data = json.load(f)

print(f"Original dataset size: {len(data)}")
print("Example entry:", data[0])
```

```python
def paraphrase_description(desc):
    templates = [
        f"A business focused on {desc}",
        f"Offering services related to {desc}",
        f"Specialized in {desc}",
        f"{desc} with a modern touch",
        f"A startup dedicated to {desc}",
        f"{desc} for modern customers",
        f"Providing innovative {desc} solutions",
        f"Experts in {desc}",
        f"Leading provider of {desc}",
        f"{desc} for everyone"
    ]
    return random.choice(templates)
```

```python
def generate_domain_variants(description):
    keywords = description.lower().split()
    base = keywords[0] if keywords else "brand"
    suffixes = [".com", ".net", ".io", ".co"]
    patterns = [
        f"{base}{random.choice(suffixes)}",
        f"get{base}{random.choice(suffixes)}",
        f"{base}hub{random.choice(suffixes)}",
        f"{base}zone{random.choice(suffixes)}",
        f"{base}{random.randint(10,99)}{random.choice(suffixes)}"
    ]
    return random.sample(patterns, 3)
```

```python
augmented_data = []

for item in data:
    # Original entry
    augmented_data.append(item)
    
    # Generate 2 paraphrased variations with new domains
    for _ in range(2):
        new_desc = paraphrase_description(item["business_description"])
        new_domains = generate_domain_variants(new_desc)
        augmented_data.append({
            "business_description": new_desc,
            "expected_domain_names": new_domains
        })

print(f"Augmented dataset size: {len(augmented_data)}")
print("Example augmented entry:", augmented_data[-1])
```

```python
output_path = "../data/augmented_dataset_v1.json"

with open(output_path, "w") as f:
    json.dump(augmented_data, f, indent=2)

print(f"Augmented dataset saved to: {output_path}")
```
